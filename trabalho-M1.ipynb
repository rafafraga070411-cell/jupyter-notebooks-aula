2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```

y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0

```

````python
import math

x1 = float(input("Insira o valor de x1: "))
y1 = float(input("Insira o valor de y1: "))
x2 = float(input("Insira o valor de x2: "))
y2 = float(input("Insira o valor de y2: "))

dx = abs(x2 - x1)
dy = abs(y2 - y1)

distancia = math.sqrt(dx**2 + dy**2)
area = math.pi * distancia**2

print()
print("x1:", x1)
print("y1:", y1)
print("x2:", x2)
print("y2:", y2)
print()
print(f"dx = {dx:.2f}")
print(f"dy = {dy:.2f}")
print(f"Distância entre os pontos: {distancia:.2f}.")
print(f"Área do círculo de raio {distancia:.2f}: {area:.2f}.")
````

### Questão 4: Encapsulamento e refatoração (1,0 ponto)

O código abaixo calcula e exibe a média de três alunos. Ele **funciona**, mas repete a mesma lógica três vezes:

```python
nota1 = 7.0
nota2 = 5.5
nota3 = 9.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Ana:", round(media, 2))

nota1 = 4.0

media = (nota1 + nota2 + nota3) / 3
print("Média de Bruno:", round(media, 2))

nota1 = 9.5

nota3 = 10.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Carla:", round(media, 2))
```

Sua tarefa:

**(a)** **Encapsule** essa lógica em uma função chamada `exibir_media(nome, nota1, nota2, nota3)`, que calcula a
média das três notas e exibe a mensagem no mesmo formato. A função deve ter uma **docstring** descrevendo o que
ela faz e o que significa cada parâmetro.

**(b)** Chame a função três vezes para reproduzir exatamente a mesma saída do código original.

**(c)** Na célula de texto indicada, explique com suas palavras: o que é **encapsulamento**, o que é
**refatoração**, e por que a versão com função é melhor que a original.

Saída esperada:

```
Média de Ana: 7.17
Média de Bruno: 5.17
Média de Carla: 9.17
```


````python
def exibir_media(nome, nota1, nota2, nota3):
    """
    Calcula a média de três notas e exibe a média do aluno.

    Parâmetros:
    nome (str): nome do aluno.
    nota1 (float): primeira nota.
    nota2 (float): segunda nota.
    nota3 (float): terceira nota.
    """
    media = (nota1 + nota2 + nota3) / 3
    print(f"Média de {nome}: {media:.2f}")

exibir_media("Ana", 7.0, 5.5, 9.0)
exibir_media("Bruno", 4.0, 6.5, 5.0)
exibir_media("Carla", 9.5, 8.0, 10.0)
````

**(c) Sua explicação sobre encapsulamento e refatoração:**

Encapsulamento é quando se junta uma parte da lógica dentro de uma função ou classe para deixar o código mais organizado, e ocultar como ela funciona, dessa forma a pessoa que usar essa parte que foi encapsulada só precisa saber como chamar e não como ela foi feita.

A refatoração é quando se melhora a estrutura do código se alterar o resultado final. Como no caso do exercício onde retirei as repetições desnecessárias e assim o código ficou mais limpo.

A versão com função é melhor pois evita repetição e facilita a leitura e manutenção do código. Pois qualquer mudança ou ajuste pode ser feito em um único lugar, ao invés de ter que mudar em cada parte repetida.

````python
from jupyturtle import *

def poligono(lados, tamanho):
    """Desenha um polígono com o
    número de lados e o tamanho do lado informados."""
    angulo = 360 / lados
    for i in range(lados):
        forward(tamanho)
        left(angulo)

make_turtle()
poligono(7, 30)
````

````python
def eh_bissexto(ano):
  return (ano % 400 == 0) or ((ano % 4 == 0) and (ano % 100 != 0))

print(1995, "é bissexto?", eh_bissexto(1995))
print(1996, "é bissexto?", eh_bissexto(1996))
print(2024, "é bissexto?", eh_bissexto(2024))
print(2025, "é bissexto?", eh_bissexto(2025))
print(2026, "é bissexto?", eh_bissexto(2026))
````

````python
def eh_triangulo(a, b, c):
  return ((c < a + b) and (b < a + c) and (a < b + c) and (a > 0 and b > 0 and c > 0))

print("eh_triangulo(6, 4, 5) ->", eh_triangulo(6, 4, 5))
print("eh_triangulo(10, 4, 5) ->", eh_triangulo(10, 4, 2))
print("eh_triangulo(-5, 6, 5) ->", eh_triangulo(-5, 6, 5))
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
def classificar_imc(peso, altura):
   """Recebe peso e altura, valida os dados, calcula o IMC e retorna a classificação."""

   if (peso <= 0) or (altura <= 0):
        return None

   imc = peso / altura ** 2

   if (imc < 18.50):
    return "Abaixo do peso"
   elif (imc >= 18.5 and imc <= 24.9):
    return "Peso normal"
   elif(imc >= 25.0 and imc <= 29.9):
    return "Sobrepeso"
   else:
    return "Obesidade"

peso = float(input("Insira o seu peso em kg:"))
altura = float(input("Insira a sua altura em metros:"))

resultado = classificar_imc(peso, altura)

if resultado  is None:
  print(f"Peso:  {peso:.2f} kg")
  print(f"Altura:  {altura:.2f} m")
  print("Dados inválidos: peso e altura devem ser positivos.")
else:
  imc = peso / altura ** 2
  print(f"Peso:  {peso:.2f} kg")
  print(f"Altura:  {altura:.2f} m")
  print(f"IMC {imc:.2f} -> {resultado}")
````

````python
def media_notas(nota1, nota2, nota3):
    """Retorna a média aritmética de três notas."""
    media = (nota1 + nota2 + nota3) / 3
    return media

def situacao(media):
    """Retorna a situação do aluno com base na média."""
    if (media >= 7):
      return "Aprovado"
    elif (media >= 4 and media < 7):
      return "Recuperação"
    else:
      return "Reprovado"

alunos = []
soma_medias = 0
aprovados = 0
melhor_nome = ""
melhor_media = -1

qtd_alunos = int(input("Insira a quantidade de alunos:"))

for i in range(qtd_alunos):
  nome = input(f"Aluno {i + 1} - nome: ")
  nota1 = float(input("Nota 1: "))
  nota2 = float(input("Nota 2: "))
  nota3 = float(input("Nota 3: "))

  media = media_notas(nota1, nota2, nota3)
  status = situacao(media)
  soma_medias += media

  if status == "Aprovado":
        aprovados += 1

  alunos.append((nome, media, status))

  if media > melhor_media:
    melhor_media = media
    melhor_nome = nome

print("=============== BOLETIM ===============")
for nome, media, status in alunos:
   print(f"{nome}: {media:.2f} - {status}")
print("=======================================")

print(f"Média geral da turma: {soma_medias / qtd_alunos:.2f}")
print(f"Aprovados: {aprovados} de {qtd_alunos} ({(aprovados / qtd_alunos) * 100:.2f}%)")
print(f"Melhor média: {melhor_nome} ({melhor_media:.2f})")
````
